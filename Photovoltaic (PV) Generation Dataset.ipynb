```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

x_cols = ['AirPressure(Pa)', 'AirTemperature(K)', 'DewPoint(K)']
y_cols = ['PV-1 Power', 'PV-2 Power', 'PV-3 Power', 'PV-4 Power', 'PV-5 Power']

fig, axes = plt.subplots(5, 3, figsize=(15, 20))

for i, y in enumerate(y_cols):
    for j, x in enumerate(x_cols):

        data = df[[x, y]].dropna()

        # Logarithmic function
        def log_func(x, a, b):
            return a * np.log(x) + b

        # Log requires x > 0
        data = data[data[x] > 0]

        popt, _ = curve_fit(log_func, data[x], data[y])

        x_line = np.linspace(data[x].min(), data[x].max(), 200)
        y_line = log_func(x_line, *popt)

        axes[i, j].scatter(data[x], data[y], s=10)
        axes[i, j].plot(x_line, y_line, color='red')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

x_cols = ['AirPressure(Pa)', 'AirTemperature(K)', 'DewPoint(K)']
y_cols = ['PV-1 Power', 'PV-2 Power', 'PV-3 Power', 'PV-4 Power', 'PV-5 Power']

fig, axes = plt.subplots(5, 3, figsize=(15, 20))

for i, y in enumerate(y_cols):
    for j, x in enumerate(x_cols):

        data = df[[x, y]].dropna()
        data = data[data[x] > 0]   # ln(x) requires x > 0

        # y = a ln(x) + b
        def log_func(x, a, b):
            return a * np.log(x) + b

        popt, _ = curve_fit(log_func, data[x], data[y])

        x_line = np.linspace(data[x].min(), data[x].max(), 200)
        y_line = log_func(x_line, *popt)

        axes[i, j].scatter(data[x], data[y], s=10)
        axes[i, j].plot(x_line, y_line, color='red')

plt.tight_layout()
plt.show()
```

```python
"""print("Baseline R2:", r2_score(y_test, p0))
print("Baseline MAE:", mean_absolute_error(y_test, p0))

print("LightGBM R2:", r2_score(y_test, p1))
print("LightGBM MAE:", mean_absolute_error(y_test, p1))"""

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
from lightgbm import LGBMRegressor

# 1. Baseline: Solar Radiation only
lr = LinearRegression()
lr.fit(X_train[["Solar Radiation"]], y_train)

p0 = lr.predict(X_test[["Solar Radiation"]])

# 2. LightGBM: all selected features
m = LGBMRegressor(
    n_estimators=400,
    learning_rate=0.05,
    random_state=0,
    n_jobs=1,
    verbosity=-1
)

m.fit(X_train, y_train)
p1 = m.predict(X_test)

# 3. Compare
print("Linear Regression R2:", r2_score(y_test, p0))
print("Linear Regression MAE:", mean_absolute_error(y_test, p0))

print("LightGBM R2:", r2_score(y_test, p1))
print("LightGBM MAE:", mean_absolute_error(y_test, p1))
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Assuming 'df' is your loaded DataFrame containing all the columns from your sample

# 1. Feature Engineering (Strictly for PV-1)
# Calculate the actual light hitting the tilted pane of PV-1
df['PV1_Effective_Irradiance'] = df['Solar Radiation'] * df['PV-1 Cosine Value of the Solar Incidence Angle']

# Thermal effect: Deviation from Standard Test Condition temperature (298.15 K)
df['PV1_Thermal_Component'] = df['PV1_Effective_Irradiance'] * (df['AirTemperature(K)'] - 298.15)

# Cooling mass effect: Air density interacting with effective solar radiation
df['PV1_Cooling_Component'] = df['PV1_Effective_Irradiance'] * df['AirDensity(kg/m3)']


# 2. Filter for Daylight Hours
# We only train the relationship using hours where the panel actually receives sun (>10 W/m²)
daylight_mask = df['PV1_Effective_Irradiance'] > 10
train_df = df[daylight_mask].copy()


# 3. Fit the Physical Relationship Matrix
if len(train_df) > 5:  # Ensure we have enough data points to fit
    features = ['PV1_Effective_Irradiance', 'PV1_Thermal_Component', 'PV1_Cooling_Component']
    X_train = train_df[features]
    y_train = train_df['PV-1 Power']
    
    # fit_intercept=False guarantees that 0 solar radiation = exactly 0 Watts generated
    model = LinearRegression(fit_intercept=False)
    model.fit(X_train, y_train)
    
    # Extract the custom mathematical constants
    coef = model.coef_
    
    print("--- PV-1 Relationship Discovered ---")
    print(f"PV-1 Optical Gain Constant (K1): {coef[0]:.6f}")
    print(f"PV-1 Thermal Degradation Loss (K2): {coef[1]:.6f} (Should be negative)")
    print(f"PV-1 Convective Cooling Gain (K3): {coef[2]:.6f}\n")
    
    # 4. Generate the Final Modeled Columns on the Full Dataset
    X_full = df[features]
    df['PV-1 Modeled Power'] = model.predict(X_full)
    
    # Enforce strict boundary conditions (No negative generation, no nighttime generation)
    df['PV-1 Modeled Power'] = df['PV-1 Modeled Power'].clip(lower=0.0)
    df.loc[df['Solar Radiation'] <= 10, 'PV-1 Modeled Power'] = 0.0
    
    # 5. Model Validation Evaluation
    y_true = train_df['PV-1 Power']
    y_pred = df.loc[daylight_mask, 'PV-1 Modeled Power']
    
    r2 = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    
    print(f"Model Performance Summary:")
    print(f"R² Score (Daylight): {r2:.4f}")
    print(f"RMSE (Daylight): {rmse:.2f} Watts")

else:
    print("Error: The current dataset slice does not contain active daylight rows where Solar Radiation > 10.")
    # Fallback initialization
    df['PV-1 Modeled Power'] = 0.0
```

```python
import pandas as pd
import numpy as np
from scipy.optimize import nnls
from sklearn.metrics import r2_score, mean_squared_error

print("==================================================")
print("   RESOLVING PHYSICAL CONSTRAINTS (K1 >= 0)      ")
print("==================================================\n")

for i in range(1, 6):
    cos_col = f'PV-{i} Cosine Value of the Solar Incidence Angle'
    actual_power_col = f'PV-{i} Power'
    
    eff_rad_col = f'PV{i}_Effective_Irradiance'
    thermal_col = f'PV{i}_Thermal_Component'
    cooling_col = f'PV{i}_Cooling_Component'
    modeled_power_col = f'PV-{i} Modeled Power'
    
    # 1. Feature Engineering
    df[eff_rad_col] = df['Solar Radiation'] * df[cos_col]
    # Invert the thermal delta so that a positive feature represents the *loss* penalty
    df[thermal_col] = df[eff_rad_col] * (df['AirTemperature(K)'] - 298.15)
    df[cooling_col] = df[eff_rad_col] * df['AirDensity(kg/m3)']
    
    # 2. Extract Daylight Mask
    daylight_mask = df[eff_rad_col] > 10
    train_df = df[daylight_mask].copy()
    
    if len(train_df) > 5:
        # To enforce K2 to be negative using a non-negative solver, 
        # we pass the negative of the thermal loss vector.
        X_train = np.column_stack((
            train_df[eff_rad_col],
            -train_df[thermal_col],  # Flips sign for the solver
            train_df[cooling_col]
        ))
        y_train = train_df[actual_power_col].values
        
        # 3. Solve with Non-Negative Constraints
        coef, residue = nnls(X_train, y_train)
        
        K1 = coef[0]
        K2 = -coef[1]  # Flip sign back to negative degradation factor
        K3 = coef[2]
        
        # 4. Generate Predictions Across Full Data Domain Range
        df[modeled_power_col] = (
            (df[eff_rad_col] * K1) + 
            (df[thermal_col] * K2) + 
            (df[cooling_col] * K3)
        )
        
        # Enforce physical boundary limits
        df[modeled_power_col] = df[modeled_power_col].clip(lower=0.0)
        df.loc[df['Solar Radiation'] <= 10, modeled_power_col] = 0.0
        
        # 5. Re-evaluate Metrics
        y_true = train_df[actual_power_col]
        y_pred = df.loc[daylight_mask, modeled_power_col]
        
        r2 = r2_score(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        
        print(f"--- [PV-{i} Physically Constrained Solution] ---")
        print(f"  • K1 (Optical Scale Factor >= 0)      : {K1:.6f}")
        print(f"  • K2 (Thermal Degradation Rate <= 0)  : {K2:.6f}")
        print(f"  • K3 (Convective Cooling Rate >= 0)   : {K3:.6f}")
        print(f"  • Metrics                             : R² = {r2:.4f} | RMSE = {rmse:.2f} W\n")

# Combine for overall balance tracking
modeled_power_cols = [f'PV-{i} Modeled Power' for i in range(1, 6)]
df['Total PV Modeled Power'] = df[modeled_power_cols].sum(axis=1)
```

```python
import pandas as pd
import numpy as np
from scipy.optimize import nnls
from sklearn.metrics import r2_score, mean_squared_error

print("==================================================")
print("   APPLYING DECOUPLED STC BASELINE NORMALIZATION  ")
print("==================================================\n")

# Standard reference density at sea level/15°C is ~1.225 kg/m3
STANDARD_DENSITY = 1.225  

for i in range(1, 6):
    cos_col = f'PV-{i} Cosine Value of the Solar Incidence Angle'
    actual_power_col = f'PV-{i} Power'
    
    eff_rad_col = f'PV{i}_Effective_Irradiance'
    thermal_col = f'PV{i}_Thermal_Adjustment'
    cooling_col = f'PV{i}_Cooling_Adjustment'
    modeled_power_col = f'PV-{i} Modeled Power'
    
    # 1. Decoupled Feature Engineering
    # Baseline driver: Clear Plane-of-Array Irradiance
    df[eff_rad_col] = df['Solar Radiation'] * df[cos_col]
    
    # Thermal impact: Irradiance modified by temperature deviation from STC (298.15K)
    df[thermal_col] = df[eff_rad_col] * (df['AirTemperature(K)'] - 298.15)
    
    # Convective cooling impact: Irradiance modified by density deviation from STC standard
    df[cooling_col] = df[eff_rad_col] * (df['AirDensity(kg/m3)'] - STANDARD_DENSITY)
    
    # 2. Extract Daylight Mask
    daylight_mask = df[eff_rad_col] > 10
    train_df = df[daylight_mask].copy()
    
    if len(train_df) > 5:
        # Enforcing physics: K1 >= 0, K2 <= 0, K3 >= 0
        # By feeding negative thermal component, NNLS forces the true K2 coefficient to be negative
        X_train = np.column_stack((
            train_df[eff_rad_col],
            -train_df[thermal_col], 
            train_df[cooling_col]
        ))
        y_train = train_df[actual_power_col].values
        
        # 3. Solve with Non-Negative Constraints
        coef, residue = nnls(X_train, y_train)
        
        K1 = coef[0]
        K2 = -coef[1]  # Re-invert to yield negative thermal decay
        K3 = coef[2]
        
        # 4. Generate Predictions Across Full Data Domain Range
        df[modeled_power_col] = (
            (df[eff_rad_col] * K1) + 
            (df[thermal_col] * K2) + 
            (df[cooling_col] * K3)
        )
        
        # Boundary constraints
        df[modeled_power_col] = df[modeled_power_col].clip(lower=0.0)
        df.loc[df['Solar Radiation'] <= 10, modeled_power_col] = 0.0
        
        # 5. Metric Validation
        y_true = train_df[actual_power_col]
        y_pred = df.loc[daylight_mask, modeled_power_col]
        
        r2 = r2_score(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        
        print(f"--- [PV-{i} Normalized Decoupled Solution] ---")
        print(f"  • K1 (True Baseline Optical Scale Factor) : {K1:.6f} (Must be > 0)")
        print(f"  • K2 (Thermal Degradation Rate)           : {K2:.6f} (Negative impact)")
        print(f"  • K3 (Convective Air Density Adjustment)  : {K3:.6f}")
        print(f"  • Metrics                                 : R² = {r2:.4f} | RMSE = {rmse:.2f} W\n")

# Combined system total balance tracking
modeled_power_cols = [f'PV-{i} Modeled Power' for i in range(1, 6)]
df['Total PV Modeled Power'] = df[modeled_power_cols].sum(axis=1)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor

print("==================================================")
print("     BUILDING PREDICTIVE OPTIMIZATION MODEL       ")
print("==================================================\n")

# 1. Prepare Features for the Machine Learning Model
features = [
    'Solar Radiation', 'AirTemperature(K)', 'AirDensity(kg/m3)', 
    'Wind Speed', 'Hour', 'Cosine of Hour', 'Sine_Hour'
]

# Target: The maximum potential power achieved by any panel orientation at that timestamp
df['Max_Observed_Panel_Power'] = df[[f'PV-{i}_Optimized_Power' for i in range(1, 6)]].max(axis=1)

# Filter for daylight hours to train efficiently
daylight_mask = df['Solar Radiation'] > 20
X_train = df.loc[daylight_mask, features]
y_train = df.loc[daylight_mask, 'Max_Observed_Panel_Power']

# 2. Train the Random Forest Optimization Model
ml_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
ml_model.fit(X_train, y_train)

# 3. Simulate an Ideal Single-Axis Tracker (FIXED THE LENGTH ALIGNMENT HERE)
# Initialize the column to 0.0 for all rows
df['PV_Total_Machine_Optimized'] = 0.0

# Only predict using daylight rows so the sizes match perfectly
predicted_daylight_peak = ml_model.predict(X_train)

# Assign the scaled predictions directly back into the daylight hours
df.loc[daylight_mask, 'PV_Total_Machine_Optimized'] = predicted_daylight_peak * 5

# 4. Calculate the Breakthrough Generation Increase
current_total_kwh = df['Total PV Power'].sum() / 1000
ml_optimized_kwh = df['PV_Total_Machine_Optimized'].sum() / 1000
net_gain_kwh = ml_optimized_kwh - current_total_kwh
percentage_gain = (net_gain_kwh / current_total_kwh) * 100

print("--- MACHINE LEARNING TRACKING UPGRADE RESULTS ---")
print(f"• Baseline Field Energy Output    : {current_total_kwh:.2f} kWh")
print(f"• ML-Optimized Tracking Output     : {ml_optimized_kwh:.2f} kWh")
print(f"• Net Energy Breakthrough Harvested : {net_gain_kwh:.2f} kWh")
print(f"• Total Field Power Increase       : +{percentage_gain:.2f}%\n")
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

print("==================================================")
print("   TRAINING & EVALUATING ML MODEL WITH VALIDATION ")
print("==================================================\n")

# 1. Feature and Target Setup
features = [
    'Solar Radiation', 'AirTemperature(K)', 'AirDensity(kg/m3)', 
    'Wind Speed', 'Hour', 'Cosine of Hour', 'Sine_Hour'
]
df['Max_Observed_Panel_Power'] = df[[f'PV-{i}_Optimized_Power' for i in range(1, 6)]].max(axis=1)

# Filter for active daylight hours where the model operates
daylight_df = df[df['Solar Radiation'] > 20].copy()

# 2. Chronological Train-Test Split (80% Train, 20% Test Validation)
split_point = int(len(daylight_df) * 0.80)

X_train = daylight_df[features].iloc[:split_point]
y_train = daylight_df['Max_Observed_Panel_Power'].iloc[:split_point]

X_test = daylight_df[features].iloc[split_point:]
y_test = daylight_df['Max_Observed_Panel_Power'].iloc[split_point:]

# 3. Train the Model on Training Split Only
ml_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
ml_model.fit(X_train, y_train)

# 4. Predict on the Unseen Test Validation Split
y_pred_test = ml_model.predict(X_test)

# 5. Compute Exact Validation Accuracy Metrics
r2 = r2_score(y_test, y_pred_test)
mae = mean_absolute_error(y_test, y_pred_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
mean_actual_power = y_test.mean()
accuracy_percentage = (1 - (mae / mean_actual_power)) * 100

print("--- MODEL VALIDATION ACCURACY METRICS ---")
print(f"• Out-of-Sample R² Score (Variance Explained) : {r2:.4f}")
print(f"• Mean Absolute Error (MAE)                  : {mae:.2f} Watts")
print(f"• Root Mean Squared Error (RMSE)              : {rmse:.2f} Watts")
print(f"• Overall Predictive Accuracy Percentage       : {accuracy_percentage:.2f}%\n")

# 6. Apply Back to the Full Dataframe for final tracking calculations
df['PV_Total_Machine_Optimized'] = 0.0
# Predict across the whole set to reconstruct your total energy totals safely
df.loc[df['Solar Radiation'] > 20, 'PV_Total_Machine_Optimized'] = ml_model.predict(daylight_df[features]) * 5
df.loc[df['Solar Radiation'] <= 20, 'PV_Total_Machine_Optimized'] = 0.0
```
